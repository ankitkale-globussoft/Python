## Real Nested Structures

```python
"""
Print all student names ✔
Print each subject mark per student ✔
Find each student’s total marks ✔
Find topper student
"""
students = [
    {
        "name": "Aman",
        "marks": {"math": 90, "science": 80}
    },
    {
        "name": "Riya",
        "marks": {"math": 95, "science": 85}
    }
]
topper_name = ""
topper_total = 0
for st in students:
    name=st["name"]
    print(name)
    marks = st["marks"]
    print(f"{name} - math: {marks['math']}, science: {marks['science']}")
    total = marks['math'] + marks['science']
    print(f"total marks: {total}")
    if total > topper_total:
        topper_name = name
        topper_total = total

print(f"Topper is {topper_name} with total marks: {topper_total}")
```

## Advanced Thinking

```python
# Grouping Problem 
data = [
    {"name": "A", "dept": "IT"},
    {"name": "B", "dept": "HR"},
    {"name": "C", "dept": "IT"}
]

grouped = {}

for item in data:
    dept = item["dept"]
    name = item["name"]

    if dept not in grouped:
        grouped[dept] = []
    
    grouped[dept] += name
print(grouped)
```

```python
# Complex nested update
"""
Increase all marks by 5
Add a new student
Find highest scorer
"""
school = {
    "class_10": {
        "students": [
            {"name": "A", "marks": 80},
            {"name": "B", "marks": 90}
        ]
    }
}
school["class_10"]["students"].append({"name": "C", "marks": 95})
highest = 0
for stu in school["class_10"]["students"]:
    name = stu['name']
    marks = stu['marks']
    marks += 5
    if marks > highest:
        highest = marks
    print(name, marks)

print(highest)
```
