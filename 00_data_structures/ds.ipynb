## Dictionary

````python
text = "bannaanaana"
char = {}
for ch in text:
    char[ch] = char.get(ch, 0) + 1
print(char)
````

# List Comprehension

## Basic Form

```python
result = [expression for item in iterable]

````python
# Create a list of squares from 1 to 10.
squares = [x*x for x in range(1, 11)]
print(squares)
````

````python
# Create a list of even numbers from 1 to 20.
even_nums = [x for x in range(1, 21) if x % 2 == 0]
# or
# even_nums = [x for x in range(2, 21, 2)]
print(even_nums)
````

````python
# Convert a list of strings to uppercase
words = ["apple", "banana", "cherry"]
ans = [s.upper() for s in words]
````

````python
# Get the length of each word.
words = ["cat", "elephant", "dog"]
ans = [len(s) for s in words]
````

````python
# Create a list containing only positive numbers.
nums = [-2, 5, -1, 7, 0]
ans = [x for x in nums if x >= 0]
````

for nested <br>
```python
[num for row in matrix for num in row]

````python
# Flatten this nested list:
matrix = [[1, 2], [3, 4], [5, 6]]
ans = [j for i in matrix for j in i]
````

````python
# Remove vowels from a string using list comprehension.
text = "list comprehension"
vovels = "aeiou"
ans = [i for i in text if i not in vovels]
````

````python
# Create a list of tuples (number, square) from 1 to 5.
ans = [(x, x*x) for x in range(1, 6)]
````

````python
# Find all numbers divisible by both 3 and 5 from 1 to 100.
ans = [x for x in range(1, 101) if x%3 == 0 and x%5 == 0]
````

````python
# Replace negative numbers with 0.
nums = [4, -3, 7, -1, 0, -8]
ans = [0 if x<0 else x for x in nums]
````

````python
# Extract names longer than 4 characters.
names = ["Ana", "Jonathan", "Mike", "Sophia"]
ans = [x for x in names if len(x) > 4]
````

````python
# Convert temperatures from Celsius to Fahrenheit.
celsius = [0, 20, 37, 100]
ans = [(x * 9/5) + 32 for x in celsius]
````

````python
# Transpose a matrix using list comprehensions.
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]

ans = [j for i in matrix for j in i]
````

````python
print(ans)
````
