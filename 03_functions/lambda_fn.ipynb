```python
# Lambda Functions
# syntax (lambda arguments: expression)

# Square of a number
sq = lambda a : a*a
print(sq(3))
print(sq(4))
print(sq(5))
```

```python
# Max of two numbers
jor = lambda a, b : a if a>b else b
print(jor(5, 22))
```

## Where Lambda is ACTUALLY used
map(lambda x: expression, iterable) <br>
filter(lambda x: expression, iterable) <br>
sorted(iterable, key=lambda x: expression)

```python
# map() — apply function to all items
# syntax:- map(lambda x: expression, iterable)

nums = [1, 2, 3, 4]
ans = list(map(lambda x: x*2, nums))
print(ans)
```

```python
# filter() - filter values
nums = [1, 2, 3, 4, 5]
ans = list(filter(lambda x: x%2==0, nums))
print(ans)
```

```python
# sorted() — sort data

students = [("Aman", 22), ("Riya", 20), ("Karan", 25)]
ans = sorted(students, key=lambda x: x[1])
print(ans)
```

```python
# Sort tuples - Sort by marks using lambda
students = [
    ("Aman", 90),
    ("Riya", 85),
    ("John", 95)
]
ans = sorted(students, key=lambda x: x[1])
print(ans)
```

```python
# Filter even numbers - Using lambda + filter()
arr = [1,2,3,4,5,6]
ans = list(filter(lambda x: x%2 == 0, arr))
print(ans)
```
