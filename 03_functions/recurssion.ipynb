```python
# Print Name N times using Recursion
def print_n(s):
    if s == 0:
        return
    print("Ankit")
    return print_n(s-1)

print_n(1)
```

```python
# Print 1 to N using Recursion
def count(n, c=1):
    if n <= 0:
        print(0)
        return
    if c == n:
        print(n)
        return
    print(c)
    return count(n, c+1)
count(10)
```

```python
# Print N to 1 using Recursion
def count(n):
    if n == 1:
        print(1)
        return
    print(n)
    return count(n-1)

count(5)
```

```python
# Sum of first N Natural Numbers
def summ(n, s=0):
    if n == 0:
        return s
    s += n
    return summ(n-1, s)
summ(6)
```

```python
# Factorial of a Number : Iterative and Recursive
def fact(n):
    if n == 1:
        return 1
    return n * fact(n-1)

fact(3)
```

```python
# Reverse a given Array
def rev_arr(a):
    p1 = 0
    p2 = len(a) - 1
    while p1 < p2:
        a[p1], a[p2] = a[p2], a[p1]
        p1 += 1
        p2 -= 1
    return a

rev_arr([1, 2, 3, 4, 5, 6])
```

```python
# Check if the given String is Palindrome or not

def is_palindrome(s):
    l = 0
    r = len(s) - 1
    while l < r:
        if s[l] != s[r]:
            return False
        l += 1
        r -= 1
    return True

is_palindrome("TAKE U FORWARD")
```

```python
# (leetcode) Valid Palindrome

class Solution:
    def isPalindrome(self, s: str) -> bool:
        s = s.lower()   
        fs=""
        for i in s:
            if i >= 'a' and i <= 'z' or i >= '0' and i <= '9':
                fs += i
        l, r = 0, len(fs)-1
        print(fs)
        while l < r:
            if fs[l] != fs[r]:
                return False
            l += 1
            r -= 1
        return True        


sl = Solution()
sl.isPalindrome("A123 man, a plan, a canal: Panam321a")
```

```python
# Print Fibonacci Series up to Nth term
def prt_fib(n):
    a, b = 0, 1
    for i in range(n+1):
        print(a)
        a, b = b, b+a

prt_fib(6)
```

```python
# (leetcode) Fibonacci Number
class Solution:
    def fib(self, n: int) -> int:
        if n == 0:
            return 0
        if n == 1:
            return 1
        return self.fib(n-1) + self.fib(n-2)
```

```python
# Count frequency of each element in the array

arr = [2,2,3,4,4,2]
fre = {}
for i in arr:
    if i not in fre:
        fre[i] = 1
    else:
        fre[i] += 1

print(fre)
```
