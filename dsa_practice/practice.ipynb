```python
n = 9

# upper part
for i in range(1, n + 1):
    spaces = " " * (n - i)
    
    if i == 1:
        print(spaces + "1")
    else:
        num = str(i) * (i - 1)
        print(spaces + num + " " + num)

# lower part
for i in range(n - 1, 0, -1):
    spaces = " " * (n - i)
    
    if i == 1:
        print(spaces + "1")
    else:
        num = str(i) * (i - 1)
        print(spaces + num + " " + num)
```

```python
n=9
for i in range(1, n+1):
    s=n-i
    if i == 1:
        print(" "*s + f"{i}", end="")
    print(" "*s + f"{i}"*(i-1) + " " + f"{i}"*(i-1))
for i in range(n-1, 0, -1):
    s=n-i
    if i == 1:
        print(" "*s + f"{i}", end="")
    print(" "*s + f"{i}"*(i-1) + " " + f"{i}"*(i-1))
```

```python
def factorial(n):
    if n == 2:
        return 2
    return n * factorial(n-1)

factorial(6)
```

```python
def power(n):
    if n == 1:
        return 2
    return (2 * power(n-1))
power(5)
```

```python
def countdown(n):
    if n == 0:
        return 0
    countdown(n-1)
    print(n)
countdown(10)
```

```python
def fibo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibo(n-1) + fibo(n-2)
```

```python
fibo(3)
```
