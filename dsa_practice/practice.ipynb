```python
def factorial(n):
    if n == 2:
        return 2
    return n * factorial(n-1)

factorial(6)
```

```python
def power(n):
    if n == 1:
        return 2
    return (2 * power(n-1))
power(5)
```

```python
def countdown(n):
    if n == 0:
        return 0
    countdown(n-1)
    print(n)
countdown(10)
```

```python
def fibo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibo(n-1) + fibo(n-2)
```

```python
fibo(3)
```

```python
def stair(n):
    if (n < 0):
        return 0
    if (n == 0):
        return 1
    
    ans = stair(n-1) + stair(n-2)
    return ans
```

```python
stair(5)
```

```python
say = ["zero", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine"]

def saydigit(num):
    if num == 0:
        return []

    last = num % 10
    num = num // 10

    return saydigit(num) + [say[last]]

saydigit(3983)
```

```python
# Reverse a String
def rs(s):
    if len(s) <= 0:
        return s
    ans = rs(s[1:])+s[0]
    return ans
rs("ankit")
```

```python
# Check Palindrome
def check(word):
    if len(word) == 0 or len(word) == 1:
        return True
    if word[0] != word[-1]:
        return False
    return check(word[1:-1])
check("w")
```

```python
def isSorted(arr):
    if len(arr) <= 1:
        return True
    if arr[0] > arr[1]:
        return False
    return isSorted(arr[1:])
```

```python
# Linear Search
def ls(arr, key, i=0):
    if i >= len(arr):
        return False
    if arr[i] == key:
        return True
    return ls(arr, key, i+1)

arr=[1, 2, 3, 4, 5]
ls(arr, 8)
```

## Binary Serach useing Recurssion

```python
def BinarySearch(arr, k):
    if len(arr) == 0:
        return False
    mid=len(arr)//2
    if arr[mid] == k:
        return True
    elif arr[mid] < k:
        return BinarySearch(arr[mid+1:], k)
    else:
        return BinarySearch(arr[:mid], k)
    
arr=[2, 4, 6, 10, 14, 16]
BinarySearch(arr, 3)
```
