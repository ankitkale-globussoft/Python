```python
# selection sort
arr = [5, 2, 9, 1]
print(arr)
for i in range(len(arr)-1):
    s = i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[s]:
            s = j
    arr[i], arr[s] = arr[s], arr[i]
print(arr)
```

```python
# bubble sort
arr = [5, 2, 9, 1]
print(arr)
for i in range(len(arr)):
    for j in range(len(arr) -1 -i):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]

print(arr)
```

```python
# insertion sort
arr = [5, 2, 9, 1]
for i in range(1, len(arr)):
    curr = arr[i]
    prev = i-1
    while prev >= 0 and arr[prev] > curr:
        arr[prev+1] = arr[prev]
        prev -= 1
    arr[prev + 1] = curr
print(arr)
```
