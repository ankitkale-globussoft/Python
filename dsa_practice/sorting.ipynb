```python
# insertion sort
arr = [5, 2, 2, 9, 1]
for i in range(1, len(arr)):
    curr = arr[i]
    prev = i-1
    while prev >= 0 and arr[prev] > curr:
        arr[prev+1] = arr[prev]
        prev -= 1
    arr[prev + 1] = curr
print(arr)
```

# Mere Sort

```python
def merge(arr1, arr2):
    l=0
    r=0
    result=[]
    
    while l < len(arr1) and r < len(arr2):
        if arr1[l] > arr2[r]:
            result.append(arr2[r])
            r+=1
        elif arr1[l] < arr2[r]:
            result.append(arr1[l])
            l+=1
        else:
            result.extend([arr1[l], arr2[r]])
            l+=1
            r+=1
    
    result.extend(arr1[l:])
    result.extend(arr2[r:])
    return result

def merge_sort(arr):
    # base
    size = len(arr)
    if size <= 1:
        return arr
    
    mid = size // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)

merge_sort([38, 27, 43, 3])
```

# Recursive Bubble Sort

```python
arr = [5, 4, 3, 2, 1]

def recur_sort(arr, i=0):
    # base case
    if i >= len(arr)-1:
        return arr
    swapped = False
    for j in range(len(arr)-i-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
            swapped = True
        
    if not swapped:
        return arr
    
    return recur_sort(arr, i+1)


recur_sort(arr)
```

# Recurssive Insertion Sort

```python
# i -> ith pos, n-> size
def recursive_insertion(arr, i, n):
    if i == n or n <= 1:
        return arr
    curr=arr[i]
    prev = i-1
    while prev >= 0 and arr[prev] > curr:
        arr[prev+1] = arr[prev]
        prev -= 1
    arr[prev+1] = curr
    return recursive_insertion(arr, i+1, n)

arr = [13,46,24,52,20,9]
recursive_insertion(arr, 1, len(arr))
```

# Quick Sort

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    mid = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + mid + quick_sort(right)

quick_sort([8, 17, 31, 32])
```

### Merge 2 sorted array

```python
arr1 = [12, 31, 35]
arr2 = [8, 17, 31, 32]
ans = []

def merge(arr1, arr2):
    if len(arr1) == 0:
        if arr2:
            ans.extend(arr2)
        return
    if len(arr2) == 0:
        if arr1:
            ans.extend(arr1)
        return
    if arr1[0] > arr2[0]:
        ans.append(arr2[0])
        return merge(arr1, arr2[1:])
    elif arr1[0] < arr2[0]:
        ans.append(arr1[0])
        return merge(arr1[1:], arr2)
    else:
        ans.extend([arr1[0], arr2[0]])
        return merge(arr1[1:], arr2[1:])
    
merge(arr1, arr2)
print(ans)
```

```python
# insertion sort
arr = [12, 11, 13, 5, 6]

def sort(arr):
    for i in range(1, len(arr)):
        curr = arr[i]
        prev = i-1
        while prev >= 0 and curr < arr[prev]:
            arr[prev+1] = arr[prev]
            prev-=1
        arr[prev+1] = curr

sort(arr)
print(arr)
```

```python
# merge sort
arr = [12, 11, 13, 5, 6]

def merge(arr1, arr2):
    i=0
    j=0
    result=[]

    while len(arr1) > i and len(arr2) > j:
        if arr1[i] >= arr2[j]:
            result.append(arr2[j])
            j+=1
        else:
            result.append(arr1[i])
            i+=1
    
    result.extend(arr1[i:])
    result.extend(arr2[j:])
    return result

def merge_sort(arr):
    size = len(arr)
    if size <= 1:
        return arr
    mid = size // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


merge_sort(arr)
```
