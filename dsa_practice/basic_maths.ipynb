```python
# Count all Digits of a Number
n = 12232345
def count(n):
    if n == 0:
        return 1
    n = abs(n)
    count=0
    while n > 0:
        count+=1
        n = n // 10
    return count

count(n)
```

```python
# Reverse a number
n = -12345
def reverse(n):
    sign = -1 if n <= 0 else 1
    rev=0
    n = abs(n)
    while n > 0:
        last = n % 10
        n = n // 10
        rev = rev*10 + last
    return sign * rev

reverse(n)
```

```python
# Reverse a number leetcode
class Solution:
    def reverse(self, x: int) -> int:
        int_max = 2147483647
        int_min = 2147483648

        sign = -1 if x <= 0 else 1
        x = abs(x)

        ans = 0

        while x > 0:
            last = x % 10
            x = x // 10
            mul = ans * 10
            if mul > int_max or (sign == -1 and mul > int_min):
                return 0
            ans = mul + last
        
        return sign * ans
    
code = Solution()
x = 1534236469
code.reverse(x)
```

```python
# Palindrome Number
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x < 0 or (x%10 == 0 and x != 0):
            return False
        rev_half = 0
        while x > rev_half:
            rev_half = x % 10 + (rev_half * 10)
            x //= 10
        
        return x == rev_half or x == rev_half // 10

code = Solution()
x = 0
code.isPalindrome(x)
```
