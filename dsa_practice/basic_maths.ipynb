```python
# Count all Digits of a Number
n = 12232345
def count(n):
    if n == 0:
        return 1
    n = abs(n)
    count=0
    while n > 0:
        count+=1
        n = n // 10
    return count

count(n)
```

```python
# Reverse a number
n = -12345
def reverse(n):
    sign = -1 if n <= 0 else 1
    rev=0
    n = abs(n)
    while n > 0:
        last = n % 10
        n = n // 10
        rev = rev*10 + last
    return sign * rev

reverse(n)
```

```python
# Reverse a number leetcode
class Solution:
    def reverse(self, x: int) -> int:
        int_max = 2147483647
        int_min = 2147483648

        sign = -1 if x <= 0 else 1
        x = abs(x)

        ans = 0

        while x > 0:
            last = x % 10
            x = x // 10
            mul = ans * 10
            if mul > int_max or (sign == -1 and mul > int_min):
                return 0
            ans = mul + last
        
        return sign * ans
    
code = Solution()
x = 1534236469
code.reverse(x)
```

```python
# Palindrome Number
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x < 0 or (x%10 == 0 and x != 0):
            return False
        rev_half = 0
        while x > rev_half:
            rev_half = x % 10 + (rev_half * 10)
            x //= 10
        
        return x == rev_half or x == rev_half // 10

code = Solution()
x = 0
code.isPalindrome(x)
```

```python
# Count Prime
# Sive of Erastothenes
def count_prime(n):
    if n < 2:
        return 0

    primes = [True] * (n + 1)
    primes[0] = primes[1] = False
    for i in range(2, int(n ** 0.5) + 1):
        if primes[i]:
            for j in range(i * i, n+1, i):
                primes[j] = False
    return sum(primes)

count_prime(100)
```

```python
class Solution:
    def countPrimes(self, n: int) -> int:
        if n < 2:
            return 0
        
        primes = [True] * n
        primes[0] = primes[1] = False

        for i in range(2, int(n ** 0.5) + 1):
            if primes[i]:
                for j in range(i * i, n, i):
                    primes[j] = False
        return sum(primes)
    
sl = Solution()
sl.countPrimes(2)
```

```python
# GCD
# Euclid's Algorithm: GCD(a, b) = GCD(b, a%b)
# now inme jab koi ek element 0 ho jayega tab jo dusra bachega voh ans hoga

def find_gcd(a, b):
    if a == 0:
        return b
    if b == 0:
        return a
    return find_gcd(b, a%b)

find_gcd(2, 10)
```

```python
# Armstrong Number
# ex: 3^3 + 7^3 + 1^3 = 371.

def is_armstrong(num):
    # sum of cubes of its digits
    n = num
    summ = 0
    while n > 0:
        summ = summ + (n % 10)**3
        n //= 10
    
    if num == summ:
        return True
    else:
        return False

is_armstrong(371)
```

```python
# Print all Divisors
def print_divisors(n):
    summ = 0
    for i in range(1, int(n ** 0.5) + 1):
        if n % i == 0:
            summ += i
            if i != n and n != n // i:
                summ += n // i
    return summ

print_divisors(1)
```

```python
# Check if a number is prime or not

def is_prime(n):
    cnt = 0
    for i in range(1, int(n ** 0.5) + 1):
        if n % i == 0:
            cnt += 1
            if i != n // i:
                cnt += 1
    return cnt == 2
```
